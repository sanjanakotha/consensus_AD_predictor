```python
import pandas as pd
import protfasta
import glob
```

# 1. Claire's sequences

Fetched 8/28/25.

```python
# Load in protein sequences from FASTA files which are in folders in fasta_file = '/global/scratch/projects/fc_mvslab/OpenProjects/Claire/transcription_factors'
fasta_files = glob.glob('/global/scratch/projects/fc_mvslab/OpenProjects/Claire/transcription_factors/*/*.fasta')
# Use protfasta one at a time per fasta file  
all_proteins = []
for fasta_file in fasta_files:
    one_protein_dict = protfasta.read_fasta(fasta_file, invalid_sequence_action = 'remove')
    one_protein_df = pd.DataFrame.from_dict(one_protein_dict, orient='index')
    all_proteins.append(one_protein_df)       
```

```python
# Full sequences
all_proteins_df = pd.concat(all_proteins)
all_proteins_df
```

```python
all_proteins_df.to_csv("../output/CL_seqs_all_8-28-25.csv")
```
