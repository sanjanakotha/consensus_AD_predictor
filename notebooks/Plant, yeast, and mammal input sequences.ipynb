# My  human TF domain search sequences

```python
mammal_TF_dict = protfasta.read_fasta("../output/interpro_uniprot_evid_1_2.fasta", invalid_sequence_action='remove')
mammal_TF_df = pd.DataFrame.from_dict(mammal_TF_dict, orient='index').reset_index()
mammal_TF_df.columns = [0, 1]
mammal_TF_df
```

```python
combined_tf_df = pd.concat([all_proteins_df, plant_TF_df, mammal_TF_df], ignore_index=True)
combined_tf_df
```

```python
# 1. Drop duplicates
combined_tf_df_unique_seqs = combined_tf_df.drop_duplicates(subset = 1)
combined_tf_df_unique_seqs
```

```python
# 2. Length filter


# Add a new column "len" to store the length of sequences in column 1
combined_tf_df_unique_seqs["len"] = combined_tf_df_unique_seqs[1].str.len()

# Set seaborn context and plot the histogram of sequence lengths
sns.set_context('talk')
sns.histplot(combined_tf_df_unique_seqs["len"])  # Plot length distribution
plt.title("Length distribution")
sns.despine()

# Calculate the 99th percentile of sequence lengths
percentile_99 = combined_tf_df_unique_seqs["len"].quantile(0.99)
print("99th percentile of sequence lengths:", percentile_99)

# Filter sequences with lengths below the 99th percentile
length_filtered = combined_tf_df_unique_seqs[combined_tf_df_unique_seqs["len"] < percentile_99]
length_filtered = length_filtered.rename(columns = {0: "id", 1 : "seq"})
length_filtered

# Uncomment the following lines if filtering sequences longer than 50 is needed
# filtered_tf_df = combined_tf_df_unique_seqs[combined_tf_df_unique_seqs["len"] > 50]
# filtered_tf_df
```

```python
# 3. Disorder filter
precalc_cl_disord = pd.read_csv("../output/cl_seqs_seq_props.csv")
display(precalc_cl_disord)
precalc_sk_disord = pd.read_csv("../output/interpro_uniprot_evid_1_2_seq_props.csv")
precalc_sk_disord
```

```python
length_filtered_with_cl_old_disord = pd.merge(length_filtered, precalc_cl_disord[["id", "percent_disorder"]], how = 'left')
length_filtered_with_cl_old_disord
```

```python
length_filtered_with_sk_disord = pd.merge(length_filtered_with_cl_old_disord, 
                                          precalc_sk_disord[["id", "percent_disorder"]], 
                                          on="id", 
                                          how="left", 
                                          suffixes=('_cl', '_sk'))
length_filtered_with_sk_disord
```

```python
length_filtered["percent_disorder"] = length_filtered_with_sk_disord["percent_disorder_cl"].combine_first(length_filtered_with_sk_disord["percent_disorder_sk"])
length_filtered
```

```python
def db_name(prefix, min_seq_id=0.5, fasta_path=False):
    if fasta_path:
        print(f"mmseqs createdb {fasta_path} {prefix}_DB") # - this alr exists
    id_thresh = int(min_seq_id * 100)
    print(f"mmseqs cluster {prefix}_DB {prefix}_DB_clu_id{id_thresh} {prefix}tmp_id{id_thresh} --min-seq-id {min_seq_id}")                      
    print(f"mmseqs createtsv {prefix}_DB {prefix}_DB {prefix}_DB_clu_id{id_thresh} {prefix}_DB_clu_id{id_thresh}.tsv")
```
