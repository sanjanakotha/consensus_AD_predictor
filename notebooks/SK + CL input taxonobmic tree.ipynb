```python
import pandas as pd
import glob 
import numpy as np
import protfasta
from ete4 import Tree
from ete4.treeview import TreeStyle, TextFace, NodeStyle
```

```python
def return_df(fasta_path):
    fasta_dict = protfasta.read_fasta(fasta_path, invalid_sequence_action='ignore')
    return pd.DataFrame({'id' : fasta_dict.keys(), 'seq' : fasta_dict.values()})
    
cl_seqs = return_df("../output/cl_seqs.fasta")
cl_seqs
```

```python
cl_seqs["species1"] = cl_seqs["id"].str.extract(r"(.{3,}_.{3,})\..*$")
cl_seqs["species2"] = cl_seqs["species1"].str.extract(r"\d_(.*)")#.str[1:]#.str[-2] + " " + cl_seqs["species"].str.split("_").str[-1]
cl_seqs["species3"] = cl_seqs["species2"].str.replace(r"_._", "_", regex=True)
cl_seqs["species4"] = cl_seqs["species3"].str.extract("_(.*)")
cl_seqs["species5"] = cl_seqs["species4"].str.extract(r"(.*)_\d")
cl_seqs["species6"] = cl_seqs["species4"].str.split("_sp").str[0]
cl_seqs["species7"] = cl_seqs["species1"].str.split("_").str[1] + "_" + cl_seqs["species1"].str.split("_").str[2]
cl_seqs["species8"] = cl_seqs["species7"].str.split("_").str[1]

cl_seqs["species1_done"] = cl_seqs["species1"].str.count("_") == 1
cl_seqs["species2_done"] = cl_seqs["species2"].str.count("_") == 1
cl_seqs["species3_done"] = cl_seqs["species3"].str.count("_") == 1
cl_seqs["species4_done"] = cl_seqs["species4"].str.count("_") == 1
cl_seqs["species5_done"] = cl_seqs["species5"].str.count("_") == 1
cl_seqs["species6_done"] = cl_seqs["species6"].str.len() > 1
cl_seqs["species7_done"] = cl_seqs["species7"].str.count("_") == 1
    
cl_seqs["done"] = cl_seqs["species1_done"] | cl_seqs["species2_done"] | cl_seqs["species3_done"] | cl_seqs["species4_done"] | cl_seqs["species5_done"] | cl_seqs["species6_done"] | cl_seqs["species7_done"]
cl_seqs
```

```python
for col in ["species1", "species2", "species3", "species4", "species5", "species6", "species7"]:
    cl_seqs[col + "_filtered"] = cl_seqs[col].apply(lambda x: x if pd.notnull(x) and x.count("_") == 1 and not any(char.isdigit() for char in x) else None)
cl_seqs
```

```python
cl_seqs["final_species"] = pd.Series([None] * len(cl_seqs), index=cl_seqs.index)
cl_seqs["final_species"] = cl_seqs["species1_filtered"].fillna(cl_seqs["species2_filtered"]).fillna(cl_seqs["species3_filtered"]).fillna(cl_seqs["species4_filtered"]).fillna(cl_seqs["species5_filtered"]).fillna(cl_seqs["species6_filtered"]).fillna(cl_seqs["species7_filtered"])
cl_seqs["final_species"] = cl_seqs["final_species"].str.split(r'\d+_').str[-1]
cl_seqs
```
